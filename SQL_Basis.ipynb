Q. What is the Difference Between a Primary Key and a Foreign Key?

->
 | Feature         | Primary Key                       | Foreign Key                                          |
| --------------- | --------------------------------- | ---------------------------------------------------- |
| Purpose         | Uniquely ID records in same table | Link tables via another table's primary key          |
| NULL values     | Never allowed                     | Allowed (optional relationships)                     |
| Duplicates      | Never allowed                     | Allowed (multiple records can reference same parent) |
| Your Q1 Example | employeeid INT PRIMARY KEY        | Could reference employeeid from another table        |
| Table limit     | 1 per table                       | Multiple possible                                    |

Q. What Are Constraints in SQL and Why Are They Used?

-> SQL Constraints are rules applied to table columns to enforce data integrity and validity. They're used to prevent bad data entry and maintain reliable database structure.

they are used for:

* Data Quality: Stop garbage data at entry

* Relationships: Enforce valid links (no orphan records)

* Consistency: Same rules across applications
